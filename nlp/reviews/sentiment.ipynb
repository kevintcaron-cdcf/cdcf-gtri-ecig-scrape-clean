```python
import pandas as pd
```

```python
df = pd.read_csv('results/getpop_sentiment_reviews.csv')
```

```python
df.columns
```

```python
# Assuming your DataFrame is named 'df'
sentiment_counts = pd.crosstab(df['Product'], df['Sentiment'])

# Optional: If you want to sort by the most reviewed products
sentiment_counts['Total'] = sentiment_counts.sum(axis=1)
sentiment_counts = sentiment_counts.sort_values(by='Total', ascending=False)
```

```python
sentiment_counts.to_csv('results/getpop_sentiment_counts.csv')
```
